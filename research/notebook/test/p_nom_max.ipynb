```python
import pypsa
```

```python
network = pypsa.Network()

network.add("Carrier", "AC")

network.add("Bus", "G1", v_nom=1, carrier="AC")
network.add("Bus", "G2", v_nom=1, carrier="AC")
network.add("Bus", "load", v_nom=1, carrier="AC")


network.add(
    "Line",
    "G1 -> load",
    bus0="G1",
    bus1="load",
    efficiency=1,
    marginal_cost=1,
    capital_cost=10,
    carrier="AC",
    length=1,
    x=0.1,
    r=0.01,
)

network.add(
    "Line",
    "G2 -> load",
    bus0="G2",
    bus1="load",
    efficiency=1,
    marginal_cost=1,
    capital_cost=10,
    carrier="AC",
    length=1,
    x=0.1,
    r=0.01,
)

network.add(
    "Generator",
    "G_G1",
    bus="G1",
    control="PQ",
    marginal_cost=1,
    capital_cost=10,
    carrier="AC",
)
network.add(
    "Generator",
    "G_G2",
    bus="G2",
    control="PQ",
    marginal_cost=1,
    capital_cost=10,
    carrier="AC",
)

network.add(
    "Load",
    "L_load",
    bus="load",
    p_set=100,
    carrier="AC",
    marginal_cost=1,
    capital_cost=10,
)

network.snapshots = [
    "2015-01-01 00:00:00",
    "2015-01-01 01:00:00",
    "2015-01-01 02:00:00",
    "2015-01-01 03:00:00",
    "2015-01-01 04:00:00",
    "2015-01-01 05:00:00",
    "2015-01-01 06:00:00",
    "2015-01-01 07:00:00",
    "2015-01-01 08:00:00",
    "2015-01-01 09:00:00",
]
```

```python
network.optimize()
```
